# Pandas

Pandas é uma biblioteca de *análise de dados* dos Python.

---

## Classes

`DataFrame`: É o coração do Pandas. É a classe que representa uma tabela.

`Series`: Segunda classe mais importante. Ela lida com listas indexadas. 

`Index`: Classe que representa índices, contando com algumas subclasses, como `RangeIndex` e `DatetimeIndex`.

`PlotAcessor`:  Raramente usada explicitamente, mas é a classe que é chamadas sempre que queremos plotar algo (lida diretamente com `matplotlib`.

---

Vamos analisar mais de perto algumas dessas classes.

### 1. Dataframe

Sempre está associado a um conjunto de índices, que não é contado entre as colunas.

#### Propriedades

- **shape:** retorna uma tupla como número de linhas e colunas do dataframe
- **dtypes:** retorna uma Series com os tipos de cada coluna
- **columns:** retorna um Index com os nomes de cada coluna
- **index:** retorna um Index com os índices do dataframe
- **T:** retorna o dataframe transposto

#### Métodos

- `head(n)`: retorna um dataframe com os $n$ primeiros resultados (o valor padrão de $n$ é $5$).
- `tail(n)`: retorna um dataframe com os $n$ últimos resultados (o valor padrão de $n$ é $5$).
- `info()`: retorna informações sobre o dataframe
- `describe()`: retorna estatísticas de cada coluna do dataframe
- `plot()`: retorna um gráfico com uma linha para cada coluna do dataframe.
  - Através do parâmetro `subplots=True`, é possível criar um gráfico separado para cada coluna 
- `set_index('coluna')`: define a coluna especificada como o conjunto de índices. É possível definir uma coluna que contém valores repetidos, mas pode dar problemas depois
- `corr()`: devolve a matriz de correlações.
  - Por padrão, usa a correlação de Pearson.
  - Usando o parâmetro `method` é possível obter também a correlação de Spearman e Kendall
- `pivot()`: transforma uma tabela long em uma tabela wide.
  - Possui os parâmetros `columns` e `values`.
- `to_numpy()`: retorna um np.ndarray do dataframe correspondente
- `isin()`
- `copy()`
- `drop_duplicates()`
- `dropna()`
- `sort_values()`

**Métodos arimtméticos:** add(), sub(), mul(), div()

**Método estatísticos:** mean(), std()

Há ainda um método especial, que é o overload dos colchetes. Isto é, `df[]`. Internamente, esse overload funciona através do dunder `__getitem__`. Dentro do colchetes é possível passar o nome de uma coluna ou uma Series de booleanos

### 2. Series

Conceitualmente, uma Series é um dataframe unidimensional. Então muitos atributos e métodos do dataframe também ocorrem numa Series.

#### Propriedades

- `index`: retorna o Index correspondente
- `values`: retorna um np.ndarray com os valores
- `name`: retorna o nome da Series
- `dtype`: retorna o tipo da Series

#### Métodos

- `head(n)`: retorna uma Series com os $n$ primeiros resultados (o valor padrão de $n$ é $5$).
- `tail(n)`: retorna uma Series com os $n$ últimos resultados (o valor padrão de $n$ é $5$).
- `info()`: retorna informações sobre a Series
- `describe()`: retorna estatísticas sobre a Series
- `plot()`: retorna um gráfico
- `corr(Series)`: retorna a correlação com a Series especificada. Novamente, por padrão usa a correlação de Pearson. Usando o parâmetro `method` é possível obter também a correlação de Spearman e Kendall
- `unique()`: retorna um np.ndarray com todos os valores distintos
- `nunique()`: retorna o número de valores distintos
- `isin(arg)`: retorna uma Series de booleanos dizendo se o valor da Series naquele índice correspondente está no arg passado

### 3. Index

É possível comparar dois índices através do método `.equals(Index)`.

## Outras funções

Há a função `pd.concat()`, que concatena dois dataframes. Através do parâmetro axis é possível especificar se a concatenação é no sentido de empilhar colunas ou de empilhar linhas. 

Tem o construtor `read_csv()` que pega um arquivo .csv e constrói um dataframe a partir dele.


# Matplotlib

## Classes

`Artist`: classe base de tudo que representa algo visualmente 

`Figure`: representa o lugar (a janela) onde os gráficos serão colocados

`Axes`: representa um gráfico. Possui `Axes3D` como subclasse. 

`Axis`: representa cada eixo de um gráfico. Possui as subclasses `XAxis` e `YAxis`.

`Tick`: objeto que representa cada tracinho + rótulo de cada eixo (possui as subclasses `XTick` e `YTick`).

`Locator`: possui diversas subclasses. Suas instâncias são responsáveis por gerenciar as posições dos ticks

`Formatter`: possui diversas subclasses. Suas instâncias são responsáveis por formatar o rótulo dos ticks

`Ticker`: contém um Locator e um Formatter

`Line2D`: representa linhas

### 1. Axis

#### Propriedades

- `major`: Ticker responsável pelos ticks maiores
- `minor`: Ticker responsável pelos ticks menores
- `label`: nome do eixo
- `offsetText`

### 2. Axes

#### Propriedades

- `xaxis`: instância de XAxis que reprsenta o eixo $x$
- `yaxis`: instância de YAxis que reprsenta o eixo $y$

#### Métodos

- `ticklabel_format()`
- `set_title()`
- `axhline()`
- `axvline()`
- `axhspan()`
- `axvspan()`

### 3. Figure

#### Métodos

- `suptitle()`
