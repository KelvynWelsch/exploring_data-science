# Regressão Linear

## Caso unidimensinal

No caso mais simples, a teoria da regressão linear se aplica a uma variável aleatória $Y$ que, por *hipótese* segue o modelo $Y = \beta_1 X + \beta_0 + \epsilon$, onde $X$ é uma variável aleatória qualquer, $\beta_0, \beta_1$ são escalares e $\epsilon$ é uma variável aleatória *gaussiana* distribuída de acordo com $\mathcal{N}(0, \sigma^2)$ e independente de $X$. 

A partir disso, colhemos $n$ realizações de $Y$, que iremos pensar simplesmente como uma realização do vetor aleatório $\mathbf{Y} = (Y_1, ..., Y_n)$, onde $Y_i = \beta_1 X_i + \beta_0 + \epsilon_i$ para cada $i \in \{1, ..., n\}$ e $(Y_i)_{i = 1}^n$ é uma família independente de variáveis aleatórias. À primeira vista, pode parecer estranho pensar em $X_i$ como uma variável aleatória já que, em muitos casos, os valores de $X_i$ são escolhidos ou dados. De qualquer modo, constantes são um caso particular de variável aleatória, então tratar como variável aleatória é, na verdade, tratar com mais generalidade. Alem do mais, mesmo nos casos em que $X_i$ não é fixo, a teoria é desenvolvida condicionalmente a $(X_1, ..., X_n)$, de modo que realmente não precisamos pensar nele como uma variável aleatória. Note que, neste caso, 

\begin{equation}
P(Y_i = y | X  = x) = \frac{1}{\sqrt{2\pi \sigma^2}} \exp\left( -\frac{(y - (\beta_1 x + \beta_0))^2}{2\sigma^2}\right).
\end{equation}

### Estimativa dos coeficientes

Nosso objetivo é usar o resultado $y = (y_1, ..., y_n)$ de uma realização de $Y$ para obter estimadores $\hat{\beta_0}$ e $\hat{\beta_1}$ de $\beta_0$ e $\beta_1$ respectivamente. Usaremos o método da *Estimativa por Máxima Verossimilhança* (MLE). Por um momento, iremos manter o valor de $\sigma^2$ fixo. De acordo com esse método, $(\hat{\beta_0}, \hat{\beta_1})$ é tal que 

\begin{equation}
 P(y_1, ..., y_n | \hat{\beta_0}, \hat{\beta_1}) = \max_{\beta_0,\beta_1} P(y_1, ..., y_n |\beta_0, \beta_1).
\end{equation}

Como cada realização é independente, 

\begin{equation}
P(y_1, ..., y_n |x_1, ..., x_n, \beta_0, \beta_1, \sigma^2) = \prod_{i = 1}^n P(y_i |x_i,\beta_0, \beta_1, \sigma^2) = \prod_{i = 1}^n \left[ \frac{1}{\sqrt{2\pi \sigma^2}} \exp\left( -\frac{(y_i - (\beta_1 x_i + \beta_0))^2}{2\sigma^2}\right) \right] = (2\pi\sigma^2)^{-n/2} \exp\left(-\sum_{i = 1}^n\frac{(y_i - (\beta_1 x_i + \beta_0))^2}{2\sigma^2} \right).
\end{equation}

Como $x \mapsto \log(x)$ é uma função estritamente crescente, o máximo de $P(y_1, ..., y_n | \beta_0, \beta_1, \sigma^2)$ é atingido exatamente no ponto de máximo de seu logaritmo:

\begin{equation}
\ell(\beta_0, \beta_1, \sigma^2) = \log(P(y_1, ..., y_n | \beta_0, \beta_1, \sigma^2)) = -\frac{n}{2}\log(2\pi) -\frac{n}{2}\log(\sigma^2) -\sum_{i = 1}^n\frac{(y_i - (\beta_1 x_i + \beta_0))^2}{2\sigma^2}.
\end{equation}

Esta função é chamada de *log-likelihood*. A menos de termos que são constante em $\beta_0$ e $\beta_1$, devemos então maximizar

\begin{equation}
\sum_{i = 1}^n (y_i - (\beta_1 x_i + \beta_0))^2.
\end{equation}

Concluímos que encontrar uma estimativa os coeficientes é equivalente a resolver um problema de mínimos quadrados. Daqui pra frente, obter os estimadores é uma conta padrão — simplesmente derivar e igual a zero. Em vez de fazer isso diretamente iremos *reescrever* a somatória acima de modo que a interpretação probabilístia e geométrica ficará mais evidente, além de facilitar a passagem para o caso multidimensional. Gostaríamos de escrever a somatória como uma norma entre vetores. Chamando novamente $(y_1, ..., y_n)$ de $y$ e $\hat{y} = (\beta_1 x_1 + \beta_0, ..., \beta_1 x_n + \beta_0)$, temos que minimizar $||y - \hat{y}||^2$, onde $||\cdot||$ denota a norma euclidiana induzida pelo produto interno canônico. Entretanto, queremos tornar a dependência de $\beta = (\beta_0, \beta_1)$ mais explícita. Se $\beta_0$ fosse $0$, podeíamos escrever simplesmente $||y - \beta_1 x||^2$, com $x = (x_1, ..., x_n)$. Apesar de este não ser o caso, podemos usar um truque para continuar vendo $\hat{y}$ como resultado da aplicação de uma transformação linear sobre $\beta$. Para tal, definimos

\begin{equation}
X =
\begin{pmatrix}
1 & x_1 \\
1 & x_2 \\
\vdots & \vdots \\
1 & x_n
\end{pmatrix},
\end{equation}

de modo que

\begin{equation}
X\beta = \begin{pmatrix}
1 & x_1 \\
1 & x_2 \\
\vdots & \vdots \\
1 & x_n
\end{pmatrix}
\begin{pmatrix}
\beta_0\\
\beta_1
\end{pmatrix} = 
\begin{pmatrix}
\beta_0 + \beta_1x_1 \\
\beta_0 + \beta_1x_2 \\
\vdots \\
\beta_0 + \beta_1x_n
\end{pmatrix}
\end{equation}.

Concluímos que o problema se resume em minimizar $||y - X\beta||^2$. Geometricamente, isso equivale a achar um vetor na imagem de $X$ que minimize a distância a $y$. Sabemos que esse vetor é a projeção ortogonal de $y$ na imagem de $X$. Temos:

\begin{align}
||y - X\beta||^2 &= (y - X\beta)^t (y - X\beta) = (y^t - \beta^t X^t)(y - X\beta) \\[0.4cm]
&= y^ty - y^tX\beta - \beta^t X^t y + \beta^t X^t X \beta
\end{align}

Uma condição necessária para enconstrarmo o ponto de mínimo é que o gradiente dessa expressão se anule. Para calcular o gradiente, comece notando algumas coisas. Em primeiro lugar, o primeiro termo é independente de $\beta$. Em segundo lugar, note que $y^t X \beta$ é um escalar, de modo que $\beta^t X^t y = (y^t X \beta)^t = y^t X \beta$. Assim, queremos achar os zeros da função $- 2\beta^t X^t y + \beta^t X^t X \beta$. Temos também que

\begin{equation}
\beta^t X^t y = (X^t y)_1 \beta_0 + (X^ty)_2 \beta_1,
\end{equation}

E portanto $\nabla_\beta (\beta^t X^t y) = X^ty$. Quanto ao último termo, temos

\begin{align}
\beta^t X^t X \beta = (\beta^t X^t X)_1 \beta_0 + (\beta^t X^t X)_2 \beta_1 &= (\beta_0 (X^t X)_{11} + \beta_1 (X^t X)_{21})\beta_0 + (\beta_0 (X^t X)_{12} + \beta_1 (X^t X)_{22})\beta_1 \\[0.4cm]
&= \beta_0^2 (X^t X)_{11} + \beta_1\beta_0 (X^t X)_{21} + \beta_0\beta_1 (X^t X)_{12} + \beta_1^2 (X^t X)_{22}.
\end{align}

Usando que $(X^tX)$ é uma matriz simétrica, temos $\beta^t X^t X \beta = \beta_0^2 (X^t X)_{11} + 2\beta_0\beta_1 (X^t X)_{12} + \beta_1^2 (X^t X)_{22}$, de modo que 

\begin{equation}
\nabla_\beta (\beta^t X^t X \beta) = (2\beta_0 (X^tX)_{11} + 2\beta_1(X^tX)_{12}, 2\beta_0(X^tX)_{21} + 2\beta_1 (X^tX)_{22}) = 2(X^tX)\beta.
\end{equation}

Juntando tudo, temos que $\hat{\beta} = (\hat{\beta_0}, \hat{\beta_1})$ deve ser tal que $-2X^ty + 2(X^tX)\hat{\beta} = 0$, isto é, $(X^tX)\hat{\beta} = X^ty$. No caso em que $(X^tX)$ é invertível, podemos achar $\hat{\beta}$ explicitamente:

$$\boxed{\hat{\beta} = (X^tX)^{-1}X^ty}$$

A matriz $X^tX$ (também conhecida como matriz de Gram) ser invertível é equivalente à transformação $X$ ser injetiva. Essa condição faz todo sentido já que, caso $X$ não seja injetiva, haverá infinitas possibilidades para $\hat{\beta}$. No nosso caso unidimensional, $X$ só nao é injetiva se todos os $x$ forem iguais.

A expressão acima obtida é um estimador para os parâmetros $\beta_0$ e $\beta_1$ dada uma certa realização de $Y$. No caso em que temos apenas o $X$ fixado, como a expressão acima vale para todo $y$, podemos escrever

$$\boxed{\hat{\beta} = (X^tX)^{-1}X^tY}$$

Neste caso, $\hat{\beta}$ se torna uma variável aleatória. Para estudar suas propriedades, vamos precisar de alguns fatos de vetores aleatórios gaussianos, então é um bom momento para pararmos e relembrarmos as propriedades sobre eles que serão usadas neste texto.

Em primeiro lugar, lembre-se de que um vetor aleatório $X$ ser gaussiano multivariado é mais forte do que pedir que cada coordenada $X_i$ seja gaussiana: o resultado de *qualquer* funcional linear deve ser gaussiano. Através do uso de funções características é fácil mostrar uma equivalência interessante para esta definição: um vetor aleatório $X$ é gaussiano multivariado se existir um vetor gaussiano padrão $Z$ (isto é, coordenadas $Z_i$ independentes e cada uma distribuídas conforme $\mathcal{N}(0, 1)$) e uma transformação afim $L$ tal que $X = L(Z)$. 

Outro fato importante é que todo vetor gaussiano multivariado $X$ em $\mathbb{R}^n$ é completamente determinado por sua média, $\mu \in \mathbb{R}^n$ e por sua matriz de covariância (ou simplesmente variância) $\Sigma \in \mathbb{R}^{n \times n}$. Este fato pode ser demonstrado novamente através do uso de funções características ou usando a definição via transformações afins. A distribuição de um vetor gaussiano multivariado com média $\mu$ e variância $\Sigma$ é denotada por $\mathcal{N}(\mu, \Sigma)$. 

É fácil provar que, se $Z \sim \mathcal{N}(\mu, \Sigma)$, então $L(Z) \sim \mathcal{N}(A\mu + b, A\Sigma A^t)$ para qualquer transformação afim da forma $L$ da forma $L(x) = Ax + b$ e $A$ linear. Disto, concluímos imediatamente que a matriz de covariância de qualquer vetor gaussiano multivariado é *positiva semi-definida*, que é equivalente a poder ser escrita como $AA^t$. É muito importante notar que $\Sigma$ não necessariamente é positiva-definida. Quando ela falha em ser positiva definida, a distribuição gaussiana multivariada é chamada de *singular*. Nos casos não-singulares, isto é, nos caso em que $\Sigma$ é de fato positiva-definida, $Z$ possui densidade de probabilidade absolutamente contínua com relação a Lebesgue em $\mathbb{R}^n$, dada por

\begin{equation}
f(x) = (2\pi)^{-n/2} (\det \Sigma)^{-1/2} \exp\left(-\frac{1}{2}(x - \mu)^t \Sigma^{-1} (x - \mu)\right),
\end{equation}

onde $\mu \in \mathbb{R}^n$ e $\Sigma$ é invertível por ser positiva-definida.

Um último fato importante sobre vetores gaussianos multivariados, que será usado, mais tarde, é que suas coordenadas são independentes se, e somente se, a matriz de covariância for diagonal. De forma geral, o fato das coordenadas de um vetor aleatório serem independentes implica matriz de covariância diagonal. A recíproca, entretanto, é uma propriedade específica de vetores gaussianos.

Apesar de não ser suficiente verificar que cada coordenada é gaussiana, essa condição se torna suficiente se também soubermos que as coordenadas formam uma família de variáveis aleatórias independentes. Isso demonstra que $Y$ é um vetor aleatório gaussiano multivariado.

Note que $Y \sim \mathcal{N}(X\beta, \sigma^2 \mathbb{1})$, de modo que automaticamente conclui-se, condicionando em $X$, que

\begin{equation}
\hat{\beta} \sim \mathcal{N}(AX\beta, AA^t \sigma^2),
\end{equation}

em que $A = (X^tX)^{-1}X^t$. Note que $A^t =  ((X^tX)^{-1}X^t)^t = X ((X^tX)^{-1})^t$. Lembrando que a inversa da transposta é a transposta da inversa, temos que $A^t= X ((X^tX)^{t})^{-1}$ $= X (X^tX)^{-1}$. Finalmente, $AA^t = (X^tX)^{-1}X^t X (X^tX)^{-1} = (X^tX)^{-1}$. Assim, condicionando em $X$,

$$\boxed{\hat{\beta} \sim \mathcal{N}(\beta, (X^tX)^{-1}\sigma^2)}$$.

Este resultado nos mostra que o valor esperado de $\hat{\beta}$ é $\beta$, isto é, este é um um estimador *não-enviesado*, com distribuição normal.

### Estimativa da variância

Nosso objetivo agora é estimar a variância $\sigma^2$. Adiantando, veremos que a estimativa por máxima verossimilhança se torna enviesada (o que não necessariamente é algo ruim, dependendo da aplicação), mas veremos estimativas alternativas que são não-enviesadas. Relembrando, a função log-likelihood é dada por

\begin{equation}
\ell(\beta_0, \beta_1, \sigma^2) = -\frac{n}{2}\log(2\pi) -\frac{n}{2}\log(\sigma^2) -\sum_{i = 1}^n\frac{(y_i - (\beta_1 x_i + \beta_0))^2}{2\sigma^2}.
\end{equation}

Derivando com relação a $\sigma^2$ e igualando a zero, chegamos na relação

\begin{equation}
-\frac{n}{2}\frac{1}{\sigma^2} + \frac{1}{2(\sigma^2)^2}\sum_{i = 1}^n[y_i - (\beta_1 x_i + \beta_0)]^2 = 0.
\end{equation}

Multiplicando por $2\sigma^2$, temos

\begin{equation}
-n + \frac{1}{\sigma^2}\sum_{i = 1}^n[y_i - (\beta_1 x_i + \beta_0)]^2 = 0,
\end{equation}

que finalmente nos dá

\begin{equation}
\sigma^2 = \frac{1}{n}\sum_{i = 1}^n[y_i - (\beta_1 x_i + \beta_0)]^2.
\end{equation}

Os termos que estão sendo elevados ao quadrado representam os *desvios* (ou *resíduos*) entre as observações e a reta determinada por $\beta$. Assim, a estimativa de $\sigma^2$, é a média do quadrado dos desvios entre o esperado e o observado. A soma em $i$ pode ser escrita como $||Y - X\beta||^2$. Note que o máximo de $\sigma^2$ depende de $\beta$. Se denotarmos o vetor de desvios por $\epsilon(\beta) = Y - X\beta$, temos $\sigma^2 = ||\epsilon(\beta)||^2/n$. Se tentarmos maximizar todas as quantidades simultaneamente, obtemos

\begin{equation}
\sigma^2 = \frac{1}{n}\sum_{i = 1}^n[y_i - (\hat{\beta_1} x_i + \hat{\beta_0})]^2 = \frac{1}{n} ||Y - X\hat{\beta}||^2 = \frac{||\hat{\epsilon}||^2}{n},
\end{equation}

onde $\hat{\epsilon}$ é uma estimativa do desvio verdadeiro $\epsilon$ associado às estimativas MLE dos parâmetros. Vamos agora analisar a distribuição de $\hat{\epsilon}$. Para tanto, precisamos falar um pouco sobre álgebra linear. Se $X$ é uma transformação linear, então a projeção ortogonal sobre a imagem de $X$ é dada por $P_X = X(X^tX)^{-1}X^t$. Tendo isto em mente, podemos escrever $X\hat{\beta} = P_XY$. Assim, $\hat{\epsilon} = Y - X\hat{\beta} = Y - P_XY = (\mathbb{1} - P_X)Y$. Note também que $(\mathbb{1} - P_X)$ é a projeção ortogonal no subespaço ortogonal à imagem de $X$. De fato, o que fizemos foi simplesmente escrever $Y$ como uma soma de uma componente paralela à imagem de $X$, a saber, $X\hat{\beta}$ e uma componente ortogonal, $\hat{\epsilon}$. 

É possível ir um pouco além lembrando que $Y = X\beta + \epsilon$, e que $(\mathbb{1} - P_X)X\beta = 0$, de modo que $\hat{\epsilon} = (\mathbb{1} - P_X)Y = (\mathbb{1} - P_X)\epsilon$. Obtemos que $\hat{\epsilon}$ é uma transformação linear do vetor gaussiano $\epsilon$ e, portanto, $\hat{\epsilon}$ também tem distribuição gaussiana multivariada. Pelo resultado que já mencionamos sobre transformações lineares de vetores gaussianos, a matriz de covariância de $\hat{\epsilon}$ é dada por $(\mathbb{1}-P_X)\sigma^2(\mathbb{1}-P_X)^t$. Lembre-se, agora, que toda projeção ortogonal é idempotente e simétrica (esses fatos podem ser facilmente deduzidos usando a fórmula explícita de $P_X$ e transmitidos para $(\mathbb{1} - P_X)$), de modo que $(\mathbb{1}-P_X)(\mathbb{1}-P_X)^t = (\mathbb{1}-P_X)(\mathbb{1}-P_X) = (\mathbb{1}-P_X)$. Concluímos que

\begin{equation}
\hat{\epsilon} \sim \mathcal{N}(0, \sigma^2 (\mathbb{1} - P_X)).
\end{equation}

Em particular, esta expressão nos diz que as componentes de $\hat{\epsilon}$ *não são* necessariamente indepedentes entre si. Assim, para analisar a distribuição de $\frac{||\hat{\epsilon}||^2}{n}$, precisamos "diagonalizar" a matriz $(\mathbb{1} - P_X)$. É um resultado clássico de álgebra linear que toda projeção ortogonal pode ser escrita como $Q^t \Sigma Q$, onde $Q$ é uma matriz ortogonal e $\Sigma$ é uma matriz diagonal com entradas $0$ e $1$ (além do mais, o número de $1$s é o posto da projeção). Tomando um tal $Q$ para nossa projeção $\mathbb{1} - P_X$, definimos $Z = Q^t\hat{\epsilon}$. Então, $Z \sim \mathcal{N}(0, \sigma^2\mathbb{1}_r)$, onde $r = n -2$ é a dimensão do subespaço ortogonal à imagem de $X$. Utilizando projeções nos subespaços de cada vetor da base que usamos para diagonalizar (e colocando primeiro os vetores que estão na imagem da projeção), temos que $Z_i \sim \mathcal{N}(0, \sigma^2)$ para $i \in \{1, ..., n-2\}$ e $Z_i \sim \mathcal{N}(0, 0) = 0$ para $i \in \{n-1, n\}$. Além do mais, as coordendas de $Z_i$ são independentes, usado um fato menciondo acima sobre vetores gaussianos com matriz de covariância diagonal.

Como $Q^t$ é ortogonal, $||Z||^2 = ||Q^t\epsilon||^2 = ||\hat{\epsilon}||^2$. Por outro lado,

\begin{equation}
||Z||^2 = \sum_{i = 1}^n Z_i^2 = \sum_{i = 1}^{n-2} Z_i ^2.
\end{equation}

Chegamos, finalmente, na conclusão que $||\hat{\epsilon}||^2$ é distribuído conforme uma soma de $n-2$ variáveis gaussianas independentes ao quadrado. Agora, notando que $Z_i/\sigma \sim \mathcal{N}(0, 1)$, temos que a quantidade

\begin{equation}
\frac{||Z||^2}{\sigma^2} = \sum_{i = 1}^{n-2} \left( \frac{Z_i}{\sigma} \right)^2
\end{equation}

é a soma do quadrado de $r$ variáveis aleatórias gaussianas padronizadas. Esse tipo de variável aleatória é importante na matemática e aparece em muitas aplicações. Tanto, que sua distribuição leva um nome especial: *chi-quadrado*, com parâmetro $r$, e denotada por $\chi^2_r$. Assim, o estimador de máxima verossimilhança pode ser escrito como

\begin{equation}
\hat{\sigma}^2_{\text{MLE}} = \frac{\sigma^2}{n}\frac{||Z||^2}{\sigma^2} \sim \frac{\sigma^2}{n}\chi^2_{n-2}.
\end{equation}
