```python
import matplotlib.pyplot as plt
import numpy as np

X = np.array([40, 60, 80, 100, 120])
beta_0 = 50
beta_1 = 3
epsilon = np.random.normal(loc=0, scale=40, size=5)
Y = beta_1*X + beta_0 + epsilon
# 4. Plotagem
plt.figure(figsize=(10, 6))

# Desenha os pontos originais (X, Y)
plt.scatter(X, Y, color='red', label='Dados Originais', zorder=5)
print(Y)
```

```python
from sklearn.linear_model import LinearRegression

# 1. Ajuste o formato de X (caso ele seja um vetor simples)
# O Scikit-learn exige que X seja 2D: [[x1], [x2], [x3]...]
X_reshaped = X.reshape(-1, 1)

# 2. Inicializa e treina o modelo
modelo = LinearRegression()
modelo.fit(X_reshaped, Y)

#print(X_reshaped)
#print(X.shape)
#print(X_reshaped.shape)

# 3. Obtém os coeficientes
a = modelo.coef_[0]   # O 'a' da reta
b = modelo.intercept_  # O 'b' da reta (onde corta o eixo Y)

print(f"Equação: y = {angular} * x + {linear}")

x_linha = np.linspace(min(X)-10, max(X)+10, 100)
y_linha = a * x_linha + b

# 4. Plotagem
plt.figure(figsize=(10, 6))

# Desenha os pontos originais (X, Y)
plt.scatter(X, Y, color='red', label='Dados Originais', zorder=5)

# Desenha a reta de regressão
plt.plot(x_linha, y_linha, color='blue', label=f'Reta: y = {a:.2f}x + {b:.2f}')
```
